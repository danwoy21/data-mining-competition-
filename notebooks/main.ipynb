## FUNCTION

```python
def preprocess_absensi(df_absensi):
    import numpy as np
    import pandas as pd

    df = df_absensi.copy()
    df['Timestamp'] = pd.to_datetime(df['Timestamp'])

    # Sort
    df = df.sort_values(by='Timestamp')

    # Fill missing Batch
    df['Kamu mengikuti Bootcamp Batch ?'] = df['Kamu mengikuti Bootcamp Batch ?'].ffill().bfill()
    df['Kamu mengikuti Bootcamp Batch ?'] = df['Kamu mengikuti Bootcamp Batch ?'].astype(float).astype(int).astype(str)

    # Ambil kolom penting dan rename
    df = df[['id', 'Timestamp', 'Kamu mengikuti Bootcamp Batch ?', 'Pertemuan ke']].rename(columns={
        'id': 'ID',
        'Kamu mengikuti Bootcamp Batch ?': 'Batch',
        'Pertemuan ke': 'Pertemuan'
    })

    # Hitung urutan per batch
    df['pertemuan_isi'] = df.groupby('Batch').cumcount().clip(upper=29) + 1
    df.loc[df['Pertemuan'].isna(), 'Pertemuan'] = df['pertemuan_isi']
    df['Pertemuan'] = df['Pertemuan'].apply(lambda x: f"Pertemuan {int(x)}" if str(x).isdigit() else x)
    df.drop('pertemuan_isi', axis=1, inplace=True)

    # Reorder & hitung ulang Pertemuan per ID + Batch
    df = df.sort_values(['ID', 'Timestamp'])
    df['Pertemuan'] = df.groupby(['ID', 'Batch']).cumcount() + 1
    df['Pertemuan'] = 'Pertemuan ' + df['Pertemuan'].astype(str)

    return df
```

```python
def hitung_ju
mlah_hadir(df_absensi_fix):
    df_jumlah_hadir = df_absensi_fix.groupby('ID').size().reset_index(name='jumlah_hadir')
    df_jumlah_hadir['persentase_hadir'] = (df_jumlah_hadir['jumlah_hadir'] / 30) * 100
    return df_jumlah_hadir
```

```python
def gabungkan_semua_fitur(df_absensi_fix, df_jumlah_hadir, df_pretest_ml, df_pretest_py, df_pretest_st, df_pendaftaran):
    df_peserta = pd.merge(df_absensi_fix, df_jumlah_hadir, on='ID', how='outer')

    # Gabung pretest
    df_pretest_ml_renamed = df_pretest_ml[['id','Score']].rename(columns={'id':'ID'})
    df_peserta = pd.merge(df_peserta, df_pretest_ml_renamed, on='ID', how='outer')
    df_peserta.rename(columns={'Score': 'score_ml'}, inplace=True)

    df_pretest_py_renamed = df_pretest_py[['id','Score']].rename(columns={'id':'ID'})
    df_peserta = pd.merge(df_peserta, df_pretest_py_renamed, on='ID', how='outer')
    df_peserta.rename(columns={'Score': 'score_py'}, inplace=True)

    df_pretest_st_renamed = df_pretest_st[['id','Score']].rename(columns={'id':'ID'})
    df_peserta = pd.merge(df_peserta, df_pretest_st_renamed, on='ID', how='outer')
    df_peserta.rename(columns={'Score': 'score_st'}, inplace=True)

    df_peserta.fillna(0, inplace=True)

    # Gabung data pendaftaran
    df_pendaftaran_renamed = df_pendaftaran[['id']].rename(columns={'id':'ID'})
    df_train_final = pd.merge(df_peserta, df_pendaftaran_renamed, on='ID', how='outer')
    df_train_final.dropna(inplace=True)

    # Convert scores to int
    for score_col in ['score_ml', 'score_py', 'score_st']:
        df_train_final[score_col] = (
            df_train_final[score_col].astype(str)
            .str.extract(r'(\d+)')
            .fillna(0)
            .astype(int)
        )

    # Hitung rata-rata
    df_train_final['rata_rata_score'] = df_train_final[['score_ml', 'score_py', 'score_st']].mean(axis=1)

    return df_train_final
```

```python
#cleaning absensi test
df_test_absensi_fix = preprocess_absensi(df_test_absensi)

# jumlah hadir test
df_jumlah_hadir_test = hitung_jumlah_hadir(df_test_absensi_fix)

#gabung semua fitur test
df_test_final = gabungkan_semua_fitur(
    df_test_absensi_fix,
    df_jumlah_hadir_test,
    df_test_pretest_ml,
    df_test_pretest_py,
    df_test_pretest_st,
    df_test_pendaftaran
)

#grouping
df_test_final_grouped = df_test_final.groupby('ID').agg({
    'jumlah_hadir': 'max',
    'persentase_hadir': 'max',
    'score_ml': 'max',
    'score_py': 'max',
    'score_st': 'max',
    'rata_rata_score': 'mean'
}).reset_index()
```
